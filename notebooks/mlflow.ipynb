````python
BASE_PATH = Path.cwd().parent
COMMANDS_FOLDER = BASE_PATH / "commands"
CACHE_PATH = BASE_PATH / "cache"
SCRIPTS_PATH = BASE_PATH / "scripts"
# FILE_PATH = CACHE_PATH / "reproduction-nc-grokking-v1_runs.parquet"
# FILE_PATH = CACHE_PATH / "canary-selection-v1_runs.parquet"
FILE_PATH = CACHE_PATH / "hyper-sweep-v1_runs.parquet"
````

````python
df = pl.read_parquet(FILE_PATH)
df.describe()
````

````python
keep_df = df.filter((pl.col("status") != "RUNNING") & pl.col("run_name").str.contains("MADD_MSE").not_())
remove_df = df.filter((pl.col("status") == "RUNNING") | pl.col("run_name").str.contains("MADD_MSE"))
````

````python
keep_df.describe()
````

````python
remove_df.describe()
````

````python
keep_path = FILE_PATH.with_name(FILE_PATH.name.replace("_runs", "_runs_keep"))
keep_df.write_parquet(keep_path)
files = [
    i for i in COMMANDS_FOLDER.iterdir()
    if i.suffix == ".txt" and i.stem[:-2] in str(FILE_PATH.stem).replace("-", "_")
]
files
````

````python
script_path = SCRIPTS_PATH / "filter_commands.py"
for f in files:
    !uv run python "{script_path}" "{f}" -r "{keep_path}"
````

````python
for id in remove_df["run_id"]:
    print(f"Removing {id}")
    client.delete_run(id)
````
